```python
from PIL import Image
import pandas as pd

img = Image.open("road.png")

sizes = {
    "원본": img.size,
    "50%": (img.width // 2, img.height // 2),
    "25%": (img.width // 4, img.height // 4)
}

modes = ["RGB", "L", "1"]

result = []

for size_name, size in sizes.items():
    resized = img.resize(size)

    for mode in modes:
        converted = resized.convert(mode)

        result.append({
            "크기": size_name,
            "해상도": f"{converted.width} x {converted.height}",
            "모드": mode,
            "메모리(Bytes)": len(converted.tobytes())
        })

df = pd.DataFrame(result)

df
```
